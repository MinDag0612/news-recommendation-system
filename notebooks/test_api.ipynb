```python
import requests


news = [
    # Sports - Football
    "Arsenal strengthen their position at the top of the Premier League after a crucial victory",

    # Sports - Basketball
    "Lakers star delivers a spectacular performance in the NBA season opener",

    # Technology
    "Apple unveils a new generation of MacBook with a faster processor",

    # Politics
    "US lawmakers debate a new bill on immigration policy",

    # Entertainment
    "Taylor Swift announces new dates for her upcoming world tour",

    # Sports - Football
    "Inter Milan move closer to the Serie A title after defeating Juventus",

    # Technology
    "Microsoft introduces new AI features for Windows users",

    # Politics
    "European leaders meet to discuss the ongoing economic crisis"
]

history = [
    # Sports - Football
    "Liverpool secure another important victory in the Premier League",
    "Manchester City remain in contention for the Premier League title",

    # Sports - Football
    "AC Milan defeat Napoli in an important Serie A match",
    "Inter Milan maintain their position near the top of Serie A",

    # Sports - Basketball
    "NBA star scores a career-high 45 points in a thrilling game",

    # Politics
    "US Congress passes a new bill after weeks of debate",

    # Entertainment
    "Hollywood star announces a new movie project"
]


# 1. Call Semantic Service
semantic_response = requests.post(
    "http://localhost:8001/semantic-encode",
    json=news,
    timeout=300
)

semantic_response.raise_for_status()

semantic_vectors = semantic_response.json()
```

```python
semantic_vectors
```

```python
# 2. Call Topic Service
topic_response = requests.post(
    "http://localhost:8002/topic-encode",
    json={
        "texts": news,
        "semantic_vectors": semantic_vectors["vectors"]
    },
    timeout=300
)

topic_response.raise_for_status()

topic_result = topic_response.json()

print(topic_result)
```

```python
recommend = requests.post(
    "http://localhost:8003/recommender-recommend",
    json={
        "news": news,
        "history": history
    },
    timeout=300
)
```

```python
recommend_result = recommend.json()
recommend_result
```
