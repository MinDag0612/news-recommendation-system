```python
import os
import pandas as pd
import sys

from pathlib import Path
current_dir = Path.cwd().resolve()
project_root = str(current_dir.parent if current_dir.name == "notebooks" else current_dir)

if project_root not in sys.path:
    sys.path.insert(0, project_root)
```

```python
from src.recommendation.RecommendationEngine import RecommendationEngine
from src.recommendation.CandidateGenerator import CandidateGenerator
from src.urv.URV import URV
from src.core.Context import VectorContext
```

```python
context = VectorContext()
behaviours = context.createHistory()
```

```python
vectors_dir = os.path.join(project_root, "vectors")
represented_vector_list = pd.read_pickle(os.path.join(vectors_dir, "represent_vectors.pkl"))

sample_behaviour = behaviours.loc[behaviours["user_id"] == "U79199"].iloc[0]
urv = URV(represented_vector=represented_vector_list)
sample_user_vector = urv.getURV(
    history=sample_behaviour["history"],
    user_id=sample_behaviour["user_id"],
    impression_id=sample_behaviour["impression_id"],
)
```

```python
candidate_generator = CandidateGenerator(represented_vector_list)

# Real recommendation mode: sample candidates from the news catalog and exclude read news.
candidate_news = candidate_generator.from_catalog(
    history=sample_behaviour["history"],
    limit=1000,
    random_state=42,
)

recommender = RecommendationEngine(represented_vector_list, alpha=0.5)
score_list = recommender.recommend(
    user_vector=sample_user_vector,
    candidate_news=candidate_news,
    k=10,
)

print(f"Generated candidates: {len(candidate_news):,}")
print(f"Top-K recommendations: {len(score_list)}")

display(score_list)
```
