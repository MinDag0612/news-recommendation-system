```python
import os
import pandas as pd
import sys

from pathlib import Path
current_dir = Path.cwd().resolve()
project_root = str(current_dir.parent if current_dir.name == "notebooks" else current_dir)

if project_root not in sys.path:
    sys.path.insert(0, project_root)
```

```python
from src.core.Models import Models
from src.core.Context import context

from src.represent.RepresentTraining import RepresentedVector
```

```python
vectors_dir = os.path.join(project_root, "vectors")
os.makedirs(vectors_dir, exist_ok=True)
vector_semantic_df = pd.read_pickle(os.path.join(vectors_dir, "semantic_vectors.pkl"))
semantic_vector_list = vector_semantic_df.set_index("News_ID")["semantic_vector"].to_dict()

vector_topic_list = pd.read_pickle(os.path.join(vectors_dir, "topic_vectors.pkl"))

title_list = context.title_list
```

```python
represented_vector = RepresentedVector(
    title_list,
    semantic_vector_list,
    vector_topic_list,
)

# print(model_context.semantic_vector_list)
# print(model_context.semantic_vector_list)
# print(model_context.topics_vector_list)

represented_vector_list = represented_vector.get_vector()
represented_vector.overview()
represented_vector.summary(sample_index=0, preview_dims=4)
```
