# Train semanric model

## Prepare title list

```python
import os
import pandas as pd
import sys

from pathlib import Path
current_dir = Path.cwd().resolve()
project_root = str(current_dir.parent if current_dir.name == "notebooks" else current_dir)

if project_root not in sys.path:
    sys.path.insert(0, project_root)
```

## Training model

```python
from src.semantic.SemanticTraining import SentenceVector
from src.core.Models import Models
from src.core.Context import context
```

```python
title_list = context.title_list
```

```python
model = Models()

sentence_vector = SentenceVector(model.sentence_model)

semantic_vector_list = sentence_vector.get_vector(
    title_list
)
# print(model_context.semantic_vector_list)
sentence_vector.overview()

metric, example = sentence_vector.summary()

print("\nSummary of Sentence Embedding:")
display(metric)
print("\nExample of Sentence Embedding:")
display(example)
```

```python
vector_df = pd.DataFrame({
    "News_ID": semantic_vector_list.keys(),
    "semantic_vector": semantic_vector_list.values()
})

vectors_dir = os.path.join(project_root, "vectors")
os.makedirs(vectors_dir, exist_ok=True)
vector_df.to_pickle(os.path.join(vectors_dir, "semantic_vectors.pkl"))
```
