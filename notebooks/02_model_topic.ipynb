```python
import os
import pandas as pd
import sys

project_root = os.path.abspath("..")   # lên 1 cấp: MIND-research

if project_root not in sys.path:
    sys.path.insert(0, project_root)
```

```python
from src.core.Models import Models
from src.core.Context import VectorContext

from src.topic.BERTopicVector import BERTopicVector
```

```python
context = VectorContext("data/primary/train_set")

title_list = context.createTitleList()

vector_df = pd.read_pickle("../vectors/primary/semantic_vectors.pkl")
semantic_vector_list = vector_df.set_index("News_ID")["semantic_vector"].to_dict()
```

```python
model = Models()

bertopic_vector = BERTopicVector(model.topic_model)

topics_vector = bertopic_vector.get_vector(
    title_list, semantic_vector_list
)
```

```python
import pickle

with open("../vectors/primary/topic_vectors.pkl", "wb") as f:
    pickle.dump(topics_vector, f)
    
bertopic_vector.model.save("../models/primary/bertopic")
```
